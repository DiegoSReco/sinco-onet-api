```python
import pandas as pd
from pathlib import Path

BASE = Path("data/processed/annotations/ruta_s/parquet")
CORRIDAS = {
    "sonnet_01": "ann-s-v1-2026-10-20-01",
    "sonnet_02": "ann-s-v1-2026-10-20-02",
    "haiku_01":  "ann-s-v1-2026-10-22-haiku-01",
    "haiku_02":  "ann-s-v1-2026-10-22-haiku-02",
    "opus_01":   "ann-s-v1-2026-10-22-opus-01",
    "opus_02":   "ann-s-v1-2026-10-22-opus-02",
}

# 1) Una tabla por corrida: tablas["opus_01"] es un DataFrame de 3542 filas
tablas = {k: pd.read_parquet(BASE / f"run_{v}.parquet") for k, v in CORRIDAS.items()}

# 2) Las seis apiladas (formato largo), con columnas "corrida" y "modelo"
largo = pd.concat([t.assign(corrida=k, modelo=k.split("_")[0]) for k, t in tablas.items()], ignore_index=True)

# 3) Formato ancho: una fila por función y una columna por corrida
ancho = largo.pivot(index=["sinco_code", "orden"], columns="corrida", values="eloundou_class")

print(largo.shape, ancho.shape)   # (21252, 32) (3542, 6)
```
